## Importing pretrained embeddings
FastText provides pretrained word embeddings for many languages, Slovene is included. Loading the model takes some time (it is 7GB large) so be patient.

```python
import fasttext
import fasttext.util
import time

start_time = time.time()
model = fasttext.load_model('cc.en.300.bin')
print("loading took %d seconds" %(time.time() - start_time))
```

## FastText object
Trained/loaded model is respresented as a <code>FastText</code> object. This offers several interesting methods:
* <code>get_dimension</code>: returns the dimension of vectors,
* <code>words</code>: returns words in the vocabulary,
* <code>get_word_vector</code>: returns the vector representing a word,
* <code>get_output_matrix</code>: returns a copy of the full input matrix of a model,
* <code>get_sentence_vector</code>: returns a single vector represenation, 
* <code>get_nearest_neighbors</code>: returns words that are most similar to the word given,
* <code>get_analogies</code>: returns word analogies ...

```python
model.get_dimension()
```

```python
print(model.words[100:200])
```

```python
print(model.get_word_vector('computer'))
```

```python
print(model.get_sentence_vector('Today it is a beautiful day.'))
```

```python
print(model.get_nearest_neighbors('mercedes', k=20))
```

```python
print(model.get_analogies('rome', 'italy', 'slovenia'))
```
